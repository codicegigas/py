Crea una función generadora leer_temperaturas() que
devuelva, una a una, tuplas con el formato: ("Ciudad",
temperatura)
- Puedes simular los datos con una lista interna de
ejemplo:
[("CDMX", 26), ("Monterrey", 34), ("Toluca", 19), ("Cancún",
38), ...]
- El generador debe tener yield y no usar return directo.

```python
def leer_temperaturas():
    """Generador que devuelve tuplas (ciudad, temperatura)."""
    datos = [
        ("CDMX", 26),
        ("Monterrey", 34),
        ("Toluca", 19),
        ("Cancún", 38),
        ("Guadalajara", 28),
        ("Puebla", 22),
    ]
    for item in datos:
        yield item


for ciudad, temp in leer_temperaturas():
        print(ciudad, temp)
```

```python
filtro = list(filter(lambda x: x[1] > 30, leer_temperaturas()))
print(filtro)
```

Convierte cada tupla a un string con el formato:
"Alerta de calor en [Ciudad]: [Temperatura]°C"

```python
conv_tupla_string = list(map(lambda x: f"Alerta de calor en {x[0]}: {x[1]}°C", filtro))
for alerta in conv_tupla_string:
    print(alerta)
```

Ordena los registros filtrados por temperatura en orden

```python
#Ordena los registros filtrados por temperatura en orden descendente.
ordenados = sorted(filtro, key=lambda x: x[1], reverse=True)
print(ordenados)
```

Calcula el promedio de temperaturas mayores a 30°C.
- Usa reduce() desde el módulo functools.

```python
from functools import reduce

suma_temperaturas = reduce(lambda acc, x: acc + x[1], filtro, 0)
promedio = suma_temperaturas / len(filtro) if filtro else 0
print(f"Temperatura promedio de temperaturas mayores a 30°C: {promedio}°C")
```

Decorador personalizado:
- Crea un decorador llamado @auditar_funcion que:
- Imprima el nombre de la función que se está
ejecutando.
- Cuente cuántas veces ha sido llamada.
- Opcional: muestre la duración de ejecución (puede
reutilizar el decorador @medir_tiempo visto en los
recursos básicos)

Salida final esperada:
- Imprimir la lista ordenada de alertas.
- Imprimir el promedio de temperaturas en formato:
"Temperatura promedio de alertas: 33.5°C"

```python
from functools import reduce
from time import perf_counter

def auditar_funcion(func):
    """Decorador que audita una función: imprime nombre, cuenta llamadas y mide duración."""
    contador = 0


    def wrapper(*args, **kwargs):
        nonlocal contador
        contador += 1
        print(f"Ejecutando función: {func.__name__} (llamada #{contador})")
        t0 = perf_counter()
        result = func(*args, **kwargs)
        t1 = perf_counter()
        dur = t1 - t0
        print(f"Duración: {dur:.6f} s")
        return result

    return wrapper


@auditar_funcion
def obtener_alertas():
    """Devuelve lista de tuplas (ciudad, temp) con temp > 30, ordenadas desc."""
    filtro = list(filter(lambda x: x[1] > 30, leer_temperaturas()))
    ordenados = sorted(filtro, key=lambda x: x[1], reverse=True)
    return ordenados


@auditar_funcion
def imprimir_alertas(alertas):
    """Imprime alertas en formato: Alerta de calor en [Ciudad]: [Temperatura]°C"""
    for ciudad, temp in alertas:
        print(f"Alerta de calor en {ciudad}: {temp}°C")


@auditar_funcion
def promedio_alertas(alertas):
    """Calcula e imprime el promedio de temperaturas de las alertas con 1 decimal."""
    
    if not alertas:
        promedio = 0
    else:
        suma_temperaturas = reduce(lambda acc, x: acc + x[1], alertas, 0)
        promedio = suma_temperaturas / len(alertas)
    print(f"Temperatura promedio de alertas: {promedio:.1f}°C")
    return promedio


# Ejecutar flujo: obtener, mostrar y calcular promedio
alertas = obtener_alertas()
print("Lista ordenada de alertas:")
imprimir_alertas(alertas)
promedio_alertas(alertas)
```
