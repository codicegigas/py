1. Filtrar clientes nuevos

A partir de dos listas: compras y registrados, obtén los nombres que aparecen en "compras", pero que no están en registrados.
Usa conjuntos y operaciones de diferencia.
Para lograr esto, se te solicita desarrollar un script en Python que use
comprensiones de listas y diccionarios. Atiende a los siguientes requisitos:

```python
compras = ["Luis", "Ana", "Luis", "Carlos", "Marta", "Ana", "Sofía", "Elena", "Luis", "Carlos"]
registrados = ["Ana", "Carlos", "Marta", "Elena"]
```

```python
from collections import ChainMap

cm = ChainMap(set(compras), set(registrados))
print(cm.maps[0] - cm.maps[1])
```

2. Eliminar duplicados y mantener orden

Obten una lista sin repeticiones de compras, conservando el orden en el que llegaron.
Usa OrderedDict o set (Justifica tu elección).

```python
from collections import OrderedDict


od = OrderedDict.fromkeys(compras)
lista_sin_repeticiones = list(od.keys())
print(lista_sin_repeticiones)

# opción explícita con set para mantener orden
compras_sin_repeticiones = []
set_compras = set()
for nombre in compras:
    if nombre not in set_compras:
        set_compras.add(nombre)
        compras_sin_repeticiones.append(nombre)
print(compras_sin_repeticiones)


# Como se observa set no cuida el orden de los elementos para ello es necesario usar OrderedDict o bien un for que nos permita recorrer la lista y agregar los elementos a un set y a una lista resultado solo si no estan en el set.
```

3. Contar cuántas veces se repite cada nombre: 

- Usa collections. Counter para contar cuántas veces aparece cada nombre en la lista compras.

```python
from collections import Counter
contador = Counter(compras)
print(contador)
```

4. Crear un resumen personalizado:

- Usando dict comprehension, crea un diccionario que relacione a cada cliente con un mensaje tipo:
- "Luis": "Ha comprado 3 veces."
- Solo incluye a quienes hayan comprado más de una vez.

```python
{nombre: f"Ha comprado {veces} veces." for nombre, veces in contador.items() if veces > 1} #Relacionamos nombre con el conteo
resumen_frecuentes = {nombre: f"Ha comprado {veces} veces." for nombre, veces in contador.items() if veces > 1} #Creamos el diccionario
print(resumen_frecuentes)
```
