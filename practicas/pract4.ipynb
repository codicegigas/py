Estás desarrollando una pequeña aplicación para el registro de usuarios,
la cual guarda los nombres, edades y correo electrónico de cada uno de
ellos. Tu sistema debe validar los datos introducidos y manejar
adecuadamente los errores que pudieran surgir, al ingresar información
incorrecta. Además, deberás incorporar mensajes de depuración para
facilitar la identificación de fallas durante el desarrollo. Sigue las
instrucciones:

Dentro del archivo, implementa una función llamada validar_datos
(nombre, edad, correo) que cumpla con los siguientes requisitos:
➔ Verifica que nombre sea una cadena de texto no vacía.
➔ Verifica que edad sea un número entero mayor que cero.
➔ Verifica que correo contenga un símbolo @.

```python
def validar_datos(nombre, edad, correo):
    """Valida los datos de un usuario.

    Reglas:
    - nombre: cadena no vacía
    - edad: entero > 0
    - correo: cadena que contiene '@'

    Lanza TypeError o ValueError según corresponda.
    Devuelve True si todo es válido.
    """
    # Verificar tipo y contenido de nombre
    if not isinstance(nombre, str):
        raise TypeError("El nombre debe ser una cadena de texto.")
    if nombre.strip() == "":
        raise ValueError("El nombre no debe estar vacío.")

    # Verificar tipo y rango de edad
    if not isinstance(edad, int):
        raise TypeError("La edad debe ser un número entero.")
    if edad <= 0:
        raise ValueError("La edad debe ser mayor que cero.")

    # Verificar tipo y formato básico de correo
    if not isinstance(correo, str):
        raise TypeError("El correo debe ser una cadena de texto.")
    if "@" not in correo:
        raise ValueError("El correo debe contener el símbolo '@'.")

    # Si todas las validaciones pasan
    return True


def registrar_usuario(nombre, edad, correo):
    """Intenta registrar un usuario usando validar_datos.

    - Captura TypeError y ValueError y los informa al usuario.
    - Garantiza que se muestre un mensaje de cierre en finally.
    - Devuelve True si se registró correctamente, False en caso contrario.
    """
    try:
        validar_datos(nombre, edad, correo)
    except (TypeError, ValueError) as e:
        print(f"Error al registrar datos: {e}")
        return False
    except Exception as e:
        # Capturar cualquier excepción inesperada
        print(f"Error inesperado: {e}")
        return False
    else:
        # Sólo si no hubo excepción
        print("Datos registrados correctamente.")
        return True
    finally:
        # Esto se ejecuta siempre
        print("Cierre: operación finalizada.")


# Ejemplos de uso / pruebas rápidas
print("-- Prueba 1: datos válidos --")
registrar_usuario("María", 28, "maria@example.com")
print()

print("-- Prueba 2: nombre vacío (ValueError) --")
registrar_usuario("   ", 28, "maria@example.com")
print()

print("-- Prueba 3: edad no entera (TypeError) --")
registrar_usuario("Juan", "veinte", "juan@example.com")
print()

print("-- Prueba 4: correo inválido (ValueError) --")
registrar_usuario("Ana", 35, "anagmail.com")
```

Implementa el uso de try, except y finally para:
➔ Capturar errores de tipo (TypeError, ValueError, etc.).
➔ Informar al usuario si los datos fueron registrados
correctamente o no.
➔ Asegurar que un mensaje de cierre siempre se muestre con
finally.

Incluye al menos 3 excepciones diferentes.
➔ Por ejemplo: ValueError, TypeError, ZeroDivisionError (si
decides extender con otra validación como división entre
edades, etc.).
• Agrega una función de prueba (probar_validaciones()) que llame a
validar_datos con diferentes entradas (válidas e inválidas),
simulando errores comunes.
• Incluye al menos dos mensajes de depuración impresos con print()
o usando el módulo logging.

```python
import logging
logging.basicConfig(level=logging.DEBUG, format="%(levelname)s: %(message)s")

def validar_datos(nombre, edad, correo):
    logging.debug(f"Validando: {nombre=}, {edad=}, {correo=}")
    if not isinstance(nombre, str):
        raise TypeError("El nombre debe ser una cadena de texto.")
    if nombre.strip() == "":
        raise ValueError("El nombre no debe estar vacío.")
    if not isinstance(edad, int):
        raise TypeError("La edad debe ser un número entero.")
    if edad < 0:
        raise ValueError("La edad no puede ser negativa.")
    if not isinstance(correo, str):
        raise TypeError("El correo debe ser una cadena.")
    if "@" not in correo:
        raise ValueError("El correo debe contener '@'.")
    logging.debug("Validación OK")
    return True

def registrar_usuario(nombre, edad, correo):
    try:
        validar_datos(nombre, edad, correo)
    except (TypeError, ValueError) as e:
        print(f"Error al registrar datos: {e}")
        return False
    except ZeroDivisionError as e:
        print(f"Error de división: {e}")
        return False
    except Exception as e:
        print(f"Error inesperado: {e}")
        return False
    else:
        print("Datos registrados correctamente.")
        return True
    finally:
        print("Cierre: operación finalizada.")
```

```python
def probar_validaciones():
    casos = [
        ("María", 28, "maria@example.com"),   # válido
        ("   ", 28, "maria@example.com"),     # ValueError: nombre vacío
        ("Juan", "veinte", "juan@example.com"), # TypeError: edad no int
        ("Ana", 0, "ana@example.com"),        # aquí simulamos ZeroDivisionError en prueba extra
    ]
    for nombre, edad, correo in casos:
        print(f"\nProbando: {nombre!r}, {edad!r}, {correo!r}")
        # caso extra: simular división por edad para lanzar ZeroDivisionError
        if isinstance(edad, int) and edad == 0:
            try:
                _ = 100 // edad   # forzamos ZeroDivisionError
            except ZeroDivisionError as e:
                print("Simulación de ZeroDivisionError:", e)
        registrar_usuario(nombre, edad, correo)

probar_validaciones()
```
